```python
# Task 5: So sánh giá trị hai chiếc pizza
import math

def pizza_unit_price(diameter_cm, price_usd):
    """
    diameter_cm: đường kính pizza (cm)
    price_usd: giá pizza (USD)
    return: giá trên mỗi mét vuông (USD/m²)
    """
    radius_cm = diameter_cm / 2
    area_cm2 = math.pi * radius_cm ** 2
    area_m2 = area_cm2 / 10000
    unit_price = price_usd / area_m2
    return unit_price


def main():
    print("So sánh giá trị của hai chiếc pizza\n")

    # Nhập thông tin pizza 1
    print("Pizza 1:")
    diameter1 = float(input("Nhập đường kính (cm): "))
    price1 = float(input("Nhập giá (USD): "))

    # Nhập thông tin pizza 2
    print("\nPizza 2:")
    diameter2 = float(input("Nhập đường kính (cm): "))
    price2 = float(input("Nhập giá (USD): "))

    # Tính giá đơn vị
    unit_price1 = pizza_unit_price(diameter1, price1)
    unit_price2 = pizza_unit_price(diameter2, price2)

    # Hiển thị kết quả
    print("\nGiá trên mỗi mét vuông:")
    print(f"Pizza 1: {unit_price1:.2f} USD/m²")
    print(f"Pizza 2: {unit_price2:.2f} USD/m²")

    # So sánh giá trị
    if unit_price1 < unit_price2:
        print("→ Pizza 1 mang lại giá trị tốt hơn.")
    elif unit_price2 < unit_price1:
        print("→ Pizza 2 mang lại giá trị tốt hơn.")
    else:
        print("→ Hai chiếc pizza có giá trị tương đương.")


# Gọi chương trình
main()
```

Task #5:  CÁCH 2

    try → thử thực hiện chương trình
    raise ValueError(...) → chủ động báo lỗi nếu dữ liệu sai
    except ValueError as e → bắt lỗi, không làm chương trình sập

```python
import math

def pizza_unit_price(d, p):
    if d <= 0 or p <= 0:
        raise ValueError("Đường kính và giá phải lớn hơn 0")
    return p / (math.pi * (d / 2) ** 2 / 10000)

try:
    # Nhập dữ liệu pizza 1
    d1 = float(input("Pizza 1 - Nhập đường kính (cm): "))
    p1 = float(input("Pizza 1 - Nhập giá (USD): "))

    # Nhập dữ liệu pizza 2
    d2 = float(input("Pizza 2 - Nhập đường kính (cm): "))
    p2 = float(input("Pizza 2 - Nhập giá (USD): "))

    # Gọi hàm tính giá đơn vị
    u1 = pizza_unit_price(d1, p1)
    u2 = pizza_unit_price(d2, p2)

    # Hiển thị kết quả
    print(f"\nPizza 1: {u1:.2f} USD/m²")
    print(f"Pizza 2: {u2:.2f} USD/m²")

    # So sánh giá trị
    if u1 < u2:
        print("→ Pizza 1 mang lại giá trị tốt hơn.")
    elif u2 < u1:
        print("→ Pizza 2 mang lại giá trị tốt hơn.")
    else:
        print("→ Hai chiếc pizza có giá trị tương đương.")

except ValueError as e:
    print("Lỗi dữ liệu:", e)
```

#Task 5 : CÁCH 3 tách các hàm ra riêng biệt để gọi hàm.
    - Hàm tính toán (pure function)

    - Hàm nhập dữ liệu

    - Hàm so sánh

```python
 import math
#khai báo hàm kiểm tra lỗi
def pizza_unit_price(diameter_cm, price_usd):
    # kiểm tra dữ liệu hợp lệ để tránh chia cho 0 / giá âm
    if diameter_cm <= 0 or price_usd <= 0:
        raise ValueError("Đường kính và giá phải lớn hơn 0.")

    area_m2 = math.pi * (diameter_cm / 2) ** 2 / 10000
    return price_usd / area_m2  # USD/m²

#khai báo hàm nhập dường kính &giá pz
def read_pizza(index):
    print(f"\nPizza {index}:")
    d = float(input("  Nhập đường kính (cm): "))
    p = float(input("  Nhập giá (USD): "))
    return d, p

#khai báo hàm so sánh giá
def so_sanh_pizza(d1, p1, d2, p2):
    u1 = pizza_unit_price(d1, p1)
    u2 = pizza_unit_price(d2, p2)

    # In ra để kiểm chứng đúng “giá trên mỗi m²”
    print(f"\nGiá đơn vị:")
    print(f"Pizza 1: {u1:.2f} USD/m²")
    print(f"Pizza 2: {u2:.2f} USD/m²")

    # So sánh: giá/m² thấp hơn => tốt hơn
    if u1 < u2:
        print("→ Pizza 1 tốt hơn (rẻ hơn theo diện tích).")
    elif u2 < u1:
        print("→ Pizza 2 tốt hơn (rẻ hơn theo diện tích).")
    else:
        print("→ Hai pizza tương đương.")

#khai báo hàm main gọi 2 hàm read_pizza và so_sanh_pizza
def main():
    try:
        d1, p1 = read_pizza(1)
        d2, p2 = read_pizza(2)
        so_sanh_pizza(d1, p1, d2, p2)
    except ValueError as e:
        print("Lỗi dữ liệu:", e)

#gọi hàm main để run
main()
```

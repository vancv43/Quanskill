**1.Lists -Danh sách:**
1️⃣ Định nghĩa: Lists (Danh sách) trong Python

List (Danh sách) trong Python là một kiểu dữ liệu có thứ tự, dùng để lưu trữ nhiều phần tử trong cùng một biến.
Các phần tử trong List có thể là cùng hoặc khác kiểu dữ liệu (số, chuỗi, boolean, object…), và có thể thay đổi được trong quá trình chương trình chạy.

📌**Đặc điểm chính của List:**

1. Có thứ tự (ordered)
2. Truy cập bằng chỉ số (index), bắt đầu từ 0
3. Thay đổi được (mutable)
4. Cho phép trùng giá trị
5.Hỗ trợ nhiều thao tác: thêm, xóa, sửa, tìm kiếm, sắp xếp, lọc…

📎**Cú pháp cơ bản:**
         my_list = [item1, item2, item3,...,n]
Ví dụ : employees = ["Nguyễn Văn An", "Trần Thị Bình", "Lê Hoàng Minh"]

**A) Thực hành đặc tính của List trong Python**
**1. Có thứ tự (ordered)**
    print("Danh sách nhân viên (theo thứ tự đã nhập):")
    print(employees)

**2. Truy cập bằng chỉ số (index), bắt đầu từ 0**
    print("Phần tử đầu tiên:", employees[0])
    print("Phần tử thứ 4:", employees[3])    # index 3
    print("Phần tử cuối:", employees[-1])   # index -1

**3. Thay đổi được (mutable)**
    print("Trước khi sửa:", employees[2])
    employees[2] = "Phạm Thị Lan (Updated)"
    print("Sau khi sửa:", employees[2])

**4. Cho phép trùng giá trị**
    employees.append("Nguyễn Văn An")  # thêm trùng tên
    print("Sau khi thêm trùng:", employees)
    print("Số lần 'Nguyễn Văn An' xuất hiện:", employees.count("Nguyễn Văn An"))

**5. Hỗ trợ nhiều thao tác: thêm, xóa, sửa, tìm kiếm, sắp xếp, lọc…**
    # Thêm
    employees.append("Ngô Minh Đức")
    print("Sau append:", employees[-3:])

    # Chèn
    employees.insert(1, "Mai Quốc Bảo")
    print("Sau insert:", employees[:3])

    # Xóa theo tên
    employees.remove("Nguyễn Văn An")  # xóa lần xuất hiện đầu tiên
    print("Sau remove:", employees[:3])

    # Xóa theo vị trí
    removed = employees.pop()  # xóa cuối
    print("Pop ra:", removed)

    # Tìm kiếm
    keyword = "Huy"
    found = [e for e in employees if keyword.lower() in e.lower()]
    print("Tìm theo từ khóa:", found)

    # Sắp xếp
    sorted_employees = sorted(employees)
    print("Sorted A-Z:", sorted_employees[:5])

    # Lọc (ví dụ: tên bắt đầu bằng 'Đ')
    filtered = [e for e in employees if e.startswith("Đ")]
    print("Lọc tên bắt đầu bằng 'Đ':", filtered)

**C) Tính lương với salaries tương ứng employees**
Cho dữ kiện: dữ liệu chi phí + bảo hiểm để tính:
    Gross = lương + phụ cấp
    Bảo hiểm nhân viên (NV): BHXH 8%, BHYT 1.5%, BHTN 1% (tổng 10.5%)
    Chi phí doanh nghiệp (DN): BHXH 17%, BHYT 3%, BHTN 1%, KPCĐ 2% (tổng 23%)

**Cho dữ liệu phụ cấp & cấu hình bảo hiểm**
# Phụ cấp theo người (triệu VND) - cùng số phần tử với employees
    allowances = [10, 8, 12, 9, 7, 6, 11, 8, 10, 9, 7, 8]  # số lượng có thể >10 do đã thêm/xóa ở phần B
    allowances = allowances[:len(employees)]  # cắt đúng bằng số NV hiện tại
    INS_EMP = {"BHXH": 0.08, "BHYT": 0.015, "BHTN": 0.01}
    INS_COM = {"KPCD": 0.02, "BHXH": 0.17, "BHYT": 0.03, "BHTN": 0.01}
    tax_rate_simple = 0.05  # mô phỏng thuế đơn giản 5% trên (gross - BH NV), để dễ thực hành

**1. Tính tổng lương (tổng salary) + tổng thu nhập gross**
    total_salary = sum(salaries)
    gross_list = [s + a for s, a in zip(salaries, allowances)]
    total_gross = sum(gross_list)

    print("Tổng lương cơ bản:", total_salary, "triệu")
    print("Tổng gross (lương + phụ cấp):", total_gross, "triệu")

**2. Tính chi phí doanh nghiệp (tổng chi phí)** 
Chi phí DN thường gồm:
    Tổng gross phải trả cho NV, phần DN phải đóng bảo hiểm (23% trên lương đóng BH; demo lấy lương cơ bản làm lương đóng BH).
    company_ins_cost = []
    for s in salaries:
        com_cost = s * sum(INS_COM.values())  # 23%
        company_ins_cost.append(com_cost)

    total_company_ins = sum(company_ins_cost)
    total_company_cost = total_gross + total_company_ins

    print("Tổng BH DN phải đóng:", round(total_company_ins, 2), "triệu")
    print("Tổng chi phí DN (Gross + BH DN):", round(total_company_cost, 2), "triệu")

**3. Trích bảo hiểm hàng loạt (nhân viên & doanh nghiệp)**
    employee_ins_deduct = []
    for s in salaries:
        emp_deduct = s * sum(INS_EMP.values())  # 10.5%
        employee_ins_deduct.append(emp_deduct)

    print("BH trích vào lương NV (10.5%) - mẫu 5 dòng đầu:")
    for i in range(min(5, len(employees))):
        print(employees[i], "->", round(employee_ins_deduct[i], 2), "triệu")

**4. Xuất “báo cáo tài chính” dạng mẫu (in ra console)**
    Ta tạo báo cáo tổng hợp kiểu doanh nghiệp:
    Tổng số NV
    Tổng lương, tổng phụ cấp, tổng gross
    BH DN, BH NV
    Thuế (mô phỏng)
    Net pay (thực lĩnh)
# Thuế mô phỏng đơn giản để thực hành
tax_list = []
net_list = []
for gross, emp_ins in zip(gross_list, employee_ins_deduct):
    taxable = max(0, gross - emp_ins)
    tax = taxable * tax_rate_simple
    net = gross - emp_ins - tax
    tax_list.append(tax)
    net_list.append(net)

report = {

    "Tổng BH NV": round(sum(employee_ins_deduct), 2),
    "Tổng thuế (mô phỏng)": round(sum(tax_list), 2),
    "Tổng thực lĩnh (Net)": round(sum(net_list), 2),
}

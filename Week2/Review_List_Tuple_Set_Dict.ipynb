ÔN TẬP DATA STRUCTURE IN PYTHON (LISTS, TUPLES, SETS, DICTIONARIES)

**NHỮNG ĐIỂM CẦN NHỚ VỀ LIST**

**A.LIST**

**1.Phòng Nhân sự quản lý danh sách nhân viên.**

employees = ["Nguyễn Văn An", "Trần Thị Bình", "Phạm Văn Cường", "Lê Thị Dung"]

salaries = [1200, 1500, 1800, 1300]  # USD

**CÂU HỎI**: 

    1.In danh sách nhân viên theo thứ tự ban đầu.

    2.Truy cập và in nhân viên đầu tiên và cuối cùng.
    
    3.Thêm nhân viên "Hoàng Minh Đức" với lương 1600.

    4.Xóa nhân viên "Trần Thị Bình" khỏi danh sách.

    5.Cập nhật lương "Phạm Văn Cường" thành 2000.

    6.Tìm nhân viên có lương cao nhất.

    7.Tính tổng lương và lương trung bình.

    8.Sắp xếp nhân viên theo lương tăng dần.

    9.Lọc danh sách nhân viên có lương ≥ 1500.

    10.Tạo danh sách mới chứa tên nhân viên viết HOA.


**NHỮNG ĐIỂM CẦN NHỚ VỀ TUPES**

**B.TUPLES – Quản lý điểm sinh viên**

student_scores = (
    ("SV001", "Nguyễn Văn An", 8.5),
    ("SV002", "Trần Thị Bình", 7.0),
    ("SV003", "Phạm Văn Cường", 9.0)
)

**Câu hỏi**


    1.In toàn bộ danh sách sinh viên.

    2. Truy cập thông tin sinh viên đầu tiên.

    3.Lấy tên sinh viên thứ 2.

    4.Giải thích vì sao tuple phù hợp cho dữ liệu điểm.

    5.Tìm sinh viên có điểm cao nhất.

    6.Tính điểm trung bình của lớp.

    7.Tách tuple thành 3 danh sách: IDs, Names, Scores.

    8.Duyệt và in sinh viên đạt điểm ≥ 8.0.

    9.Chuyển tuple sang list để bổ sung sinh viên mới.

    10.So sánh Tuple vs List trong hệ thống quản lý học tập.

**NHỮNG ĐIỂM CẦN NHỚ VỀ SETS**

**C – SET (Tập hợp) – Phân tích khách hàng**
Phân loại khách hàng theo hành vi mua sắm.

online_customers = {"An", "Bình", "Cường", "Dung"}

offline_customers = {"Bình", "Dung", "Hà", "Minh"}

**Câu hỏi**

    1.In danh sách khách hàng mua online.

    2.Tìm khách hàng mua cả online và offline.

    3.Tìm khách hàng chỉ mua online.

    4.Tìm khách hàng chỉ mua offline.

    5.Tạo tập hợp khách hàng duy nhất.

    6.Kiểm tra "An" có phải khách hàng online không.

    7.Thêm khách hàng "Lan" vào tập online.

    8.Xóa "Cường" khỏi tập online.

    9.Giải thích vì sao set không cho phép trùng lặp.

    10.Ứng dụng set trong loại bỏ dữ liệu trùng khi thu thập data.


**NHỮNG ĐIỂM CẦN NHỚ VỀ DICTIONARIES**

**D – DICTIONARY (Từ điển) – Quản lý sản phẩm**

Quản lý Hệ thống quản lý sản phẩm thương mại điện tử.

    products = {
    "P001": {"name": "Laptop", "price": 1200, "stock": 10},
    "P002": {"name": "Mouse", "price": 25, "stock": 100},
    "P003": {"name": "Keyboard", "price": 45, "stock": 50}
}

**Câu hỏi**

    1.In toàn bộ danh sách sản phẩm.

    2.Truy cập thông tin sản phẩm "P002".

    3.Cập nhật giá "Laptop" thành 1100.

    4.Giảm tồn kho "Mouse" đi 10 sản phẩm.

    5.Thêm sản phẩm mới "P004" – Monitor.

    6.Xóa sản phẩm "P003".

    7.Tính tổng giá trị tồn kho.

    8.Tìm sản phẩm có giá cao nhất.

    9.Lọc sản phẩm có tồn kho < 20.

    10.Giải thích vì sao Dictionary phù hợp cho hệ thống CRUD.

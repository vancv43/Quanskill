**1.Định nghĩa Sets – nhìn từ góc độ Data Science**
    Khái niệm Set trong Python là một collection có 4 đặc trưng quan trọng:
    | Thuộc tính                  | Ý nghĩa                   |
| --------------------------- | ------------------------- |
| 1_Không có thứ tự           | Không truy cập bằng index |
| 2_Không cho phép duplicate  | Tự động loại bỏ trùng     |
| 3_Có thể thay đổi (mutable) | Add / remove phần tử      |
| 4_Tối ưu cho membership     | `in`, `not in` rất nhanh  |

**So sánh List, Tuple, Set and Dict** 
| Kiểu dữ liệu | Có thứ tự | Cho trùng | Dùng khi nào        |
| ------------ | --------- | --------- | ------------------- |
| `list`       | có (yes)  | có (yes)  | Lưu chuỗi sự kiện   |
| `tuple`      | có (yes)  | có(yes)   | Dữ liệu cố định     |
| `set`        |  không(no)|❌không(no)| Loại trùng, so sánh |
| `dict`       |  không(no)| (không) key| Ánh xạ – tra cứu   |

**2.3. Các hàm kiểm tra quan hệ tập hợp**
    A = {1, 2}          # A có 2 phần tử: 1, 2
    B = {1, 2, 3, 4}    # B có 4 phần tử: 1, 2, 3, 4

    A.issubset(B)      # Kiểm tra A có phải là tập con của B không? -> Kết quả trả tra là True 
    B.issuperset(A)    # Kiểm tra B có phải là tập cha (tập bao) của A không? ->Kết quả trả tra là True 
    A.isdisjoint({5})  # Kiểm tra hai tập có KHÔNG CÓ phần tử chung hay không ->Kết quả trả tra là True 

| Hàm            | Câu hỏi trả lời           | True khi |
| -------------- | ------------------------- | -------- |
| issubset()   | A có nằm trong B?         | A ⊆ B    |
| issuperset() | B có bao A?               | B ⊇ A    |
| isdisjoint() | Hai tập có chung phần tử? | Giao = ∅ |

```python
#2. Thao tác với Set - Website visitors
monday_visitors = {"user1", "user2", "user3", "user4"}
tuesday_visitors = {"user3", "user4", "user5", "user6"}

# Add and remove elements
monday_visitors.add("user7")
print(monday_visitors)
# Output: {'user1', 'user2', 'user3', 'user4', 'user7'}

monday_visitors.remove("user1")   # Raises error if not found
print(monday_visitors)
# Output: {'user2', 'user3', 'user4', 'user7'}

monday_visitors.discard("user999")  # No error if not found
print(monday_visitors)
# Output: {'user2', 'user3', 'user4', 'user7'}

# Check membership
print("user2" in monday_visitors)   # Output: True
print("user5" in monday_visitors)   # Output: False

# Set length
print(len(monday_visitors))         # Output: 4

#len : đếm số phần tử trong set/list/dict /tuple
```

```python
# Customer segments
premium_customers = {"Alice", "Bob", "Charlie", "David"}
active_customers = {"Bob", "Diana", "Eve", "Charlie"}

# Union - all customers
all_customers = premium_customers | active_customers
print(all_customers)
# Output: {'Alice', 'Bob', 'Charlie', 'David', 'Diana', 'Eve'}

# Intersection - customers in both segments
both_segments = premium_customers & active_customers
print(both_segments)
# Output: {'Bob', 'Charlie'}

# Difference - premium customers who are not active
premium_only = premium_customers - active_customers
print(premium_only)
# Output: {'Alice', 'David'}

# Symmetric difference - customers in only one segment
exclusive_customers = premium_customers ^ active_customers
print(exclusive_customers)
# Output: {'Alice', 'David', 'Diana', 'Eve'}
```

```python
# =========================
# HOMEWORK 2 - Full Python Code
# Topic: Customer engagement analysis using Sets + business logic
# Rule: "Highly Engaged" = appears in AT LEAST 2 of 3 sets:
#       (email_subscribers, purchase_customers, mobile_app_users)
# =========================

# --- Input data ---
email_subscribers = {"cust1", "cust2", "cust3", "cust4", "cust5"}
purchase_customers = {"cust2", "cust4", "cust6", "cust7", "cust8"}
mobile_app_users = {"cust1", "cust3", "cust6", "cust9", "cust10"}

print("=== INPUT DATA ===")
print("email_subscribers :", email_subscribers)
print("purchase_customers:", purchase_customers)
print("mobile_app_users  :", mobile_app_users)
print()

# -------------------------
# CÂU 1: Pairwise intersections
# -------------------------
email_and_purchase = email_subscribers & purchase_customers
email_and_app = email_subscribers & mobile_app_users
purchase_and_app = purchase_customers & mobile_app_users

print("=== CÂU 1: Pairwise intersections ===")
print("email ∩ purchase  =", email_and_purchase, "| count =", len(email_and_purchase))
print("email ∩ app       =", email_and_app,      "| count =", len(email_and_app))
print("purchase ∩ app    =", purchase_and_app,   "| count =", len(purchase_and_app))
print()

# -------------------------
# CÂU 2: Highly engaged = in at least 2 sets
# Logic: (email∩purchase) ∪ (email∩app) ∪ (purchase∩app)
# -------------------------
highly_engaged_customers = (email_and_purchase | email_and_app | purchase_and_app)

print("=== CÂU 2: Highly engaged customers (>=2 channels) ===")
print("highly_engaged_customers =", highly_engaged_customers)
print("count =", len(highly_engaged_customers))
print()

# -------------------------
# CÂU 3: Compare counts
# - Total unique customers across 3 channels
# - Customers appearing in exactly 1 channel
# -------------------------
total_customers = email_subscribers | purchase_customers | mobile_app_users

# Exactly 1 channel:
# Option A (clear, explicit):
only_email = email_subscribers - (purchase_customers | mobile_app_users)
only_purchase = purchase_customers - (email_subscribers | mobile_app_users)
only_app = mobile_app_users - (email_subscribers | purchase_customers)
exactly_one_channel = only_email | only_purchase | only_app

print("=== CÂU 3: Comparison ===")
print("total_customers =", total_customers)
print("total count     =", len(total_customers))
print()

print("only_email      =", only_email,     "| count =", len(only_email))
print("only_purchase   =", only_purchase,  "| count =", len(only_purchase))
print("only_app        =", only_app,       "| count =", len(only_app))
print()

print("exactly_one_channel =", exactly_one_channel)
print("exactly_one_channel count =", len(exactly_one_channel))
print()

# -------------------------
# CÂU 4 (Bonus): subset check
# -------------------------
is_subset = highly_engaged_customers.issubset(total_customers)

print("=== CÂU 4 (Bonus): Subset check ===")
print("highly_engaged_customers ⊆ total_customers ?", is_subset)

# Optional explanation printed
if is_subset:
    print("Explanation: All highly engaged customers are part of the total customer universe (union of all channels).")
else:
    print("Explanation: Something is wrong: a 'highly engaged' customer should always appear in total_customers.")
```

**1.Định nghĩa Dictionaries**
Dictionary là cấu trúc dữ liệu lưu trữ dữ liệu dưới dạng key–value pairs,
    Có thứ tự (từ Python 3.7+)
    Có thể thay đổi (mutable)

    Value có thể là bất kỳ kiểu dữ liệu nào

**Dictionary được dùng để:**
    Tra cứu dữ liệu nhanh
    Biểu diễn đối tượng thực tế (student, customer, product, employee)
    Nền tảng của JSON, API, Data Science


```python
# Create empty dictionary
empty_dict = {}
print("This is an empty dictionary:", empty_dict)   # Output: {}

# Create dictionary with data
student_grades = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
    "Diana": 96
}
print(student_grades)
# Output: {'Alice': 85, 'Bob': 92, 'Charlie': 78, 'Diana': 96}

# Using dict() constructor
product_prices = dict(laptop=999, mouse=25, keyboard=75)
print(product_prices)
# Output: {'laptop': 999, 'mouse': 25, 'keyboard': 75}

#Giải thích: Keys phải unique và immutable (string, number, tuple).
##Values có thể là bất kỳ data type nào.
```

**Truy cập và thay đổi dữ liệu**

```python
# Sales data by region
sales_by_region = {
    "North": 15000,
    "South": 12000,
    "East": 18000,
    "West": 14000
}

# Access values by key
print(sales_by_region["North"])           # Output: 15000
print(sales_by_region.get("North"))       # Output: 15000
print(sales_by_region.get("Central", 0))  # Output: 0 (default value)

# Modify existing value
sales_by_region["North"] = 16000
print(sales_by_region["North"])           # Output: 16000

# Add new key-value pair
sales_by_region["Central"] = 10000
print(sales_by_region)

# Remove key-value pair
del sales_by_region["Central"]
removed_value = sales_by_region.pop("West", 0)
print(f"Removed West sales: {removed_value}")
# Output: Removed West sales: 14000

```

```python
# Product sales data
sales_data = {
    "products": ["laptop", "mouse", "keyboard", "monitor"],
    "quantities": [50, 200, 150, 75],
    "prices": [999, 25, 75, 299],
    "categories": ["electronics", "accessories", "accessories", "electronics"]
}

# Calculate total revenue by product
for i in range(len(sales_data["products"])):
    product = sales_data["products"][i]
    revenue = sales_data["quantities"][i] * sales_data["prices"][i]
    print(f"{product}: ${revenue}")

# Count products by category
category_count = {}
for category in sales_data["categories"]:
    category_count[category] = category_count.get(category, 0) + 1

print(category_count)
# Output: {'electronics': 2, 'accessories': 2}
```

```python
# Customer transaction data
customers = {
    "customer001": {
        "name": "Alice Johnson",
        "transactions": [150, 200, 75, 300],   # List of amounts
        "location": (40.7128, -74.0060),       # Tuple for coordinates
        "purchase_categories": {"electronics", "clothing", "books"}  # Set
    },
    "customer002": {
        "name": "Bob Smith",
        "transactions": [100, 250, 180],
        "location": (34.0522, -118.2437),
        "purchase_categories": {"electronics", "sports", "books"}
    }
}

# Analysis
for customer_id, data in customers.items():
    name = data["name"]
    total_spent = sum(data["transactions"])          # List operation
    avg_transaction = total_spent / len(data["transactions"])
    lat, lon = data["location"]                      # Tuple unpacking
    num_categories = len(data["purchase_categories"])  # Set size

    print(f"Customer: {name}")
    print(f"Total spent: ${total_spent}, Average: ${avg_transaction:.2f}")
    print(f"Location: ({lat}, {lon}), Categories: {num_categories}")
    print("---")
```

**1. Định nghĩa Tuples (Bộ dữ liệu)**
**1.1. Tuples là gì?**

Tuple là một cấu trúc dữ liệu trong Python:
    Có thứ tự (ordered)
    Không thay đổi được (immutable)
    Cho phép trùng giá trị
    Có thể chứa nhiều kiểu dữ liệu khác nhau

**Ví dụ :**
1/ Quản lý nhân viên (HR): 
employee = ("E001", "Nguyen Van A", "IT", 1500)
2/ Tuple cho Log hệ thống (IT / DevOps) 
system_log = ("2026-01-09 14:32:10", "ERROR", "AuthService", "Invalid token")
| Phần tử   | Nội dung      |
| --------- | ------------- |
| Thời gian | Timestamp     |
| Mức độ    | ERROR         |
| Service   | AuthService   |
| Thông báo | Invalid token |

3/ Tuple cho Giao dịch tài chính (Finance / Banking): 
transaction = ("TXN20260109", "E001", 2500.75, "USD", "SUCCESS")

4/ Tuple cho Feature vector trong Data Science / AI: 
feature_vector = (0.82, 1.35, 0.12, 0.67)


**Bài tập 1: Quản lý nhân viên (HR)**
Yêu cầu: In ra nhân viên "Nguyễn Thành Luân" làm việc ở phòng IT và có mức lương 1500.

```python
employee = ("E001", "Nguyễn Thành Luân", "IT", 1500)

emp_id, name, dept, salary = employee #employee unpacking 

print(f"Employee: {name} works in {dept}, salary = {salary}") # f: formatted string
```

**Bài tập 2:Tuple cho Log hệ thống (IT / DevOps)**
Cho system_log = ("2026-01-09 14:32:10", "ERROR", "AuthService", "Invalid token")

| Phần tử   | Nội dung      |
| --------- | ------------- |
| Thời gian | Timestamp     |
| Mức độ    | ERROR         |
| Service   | AuthService   |
| Thông báo | Invalid token |
Yêu cầu: 
Dùng tuple unpacking để gán các giá trị vào 4 biến:
timestamp, level, service, message
In ra log theo định dạng: [ERROR] AuthService at 2026-01-09 14:32:10: Invalid token

```python
# Tuple system log
system_log = ("2026-01-09 14:32:10", "ERROR", "AuthService", "Invalid token")

# Tuple unpacking
timestamp, level, service, message = system_log

# In log theo định dạng yêu cầu
print(f"[{level}] {service} at {timestamp}: {message}")
```

**BÀI TẬP 5: KẾT HỢP LIST & TUPLE TRONG DOANH NGHIỆP** 

Công ty lưu trữ danh sách nhân viên, trong đó:
    - Mỗi nhân viên là một tuple (dữ liệu gốc, không thay đổi)
    - Toàn bộ danh sách là một list (có thể thêm / xóa). 
Employees = [
    ("E001", "Nguyen Van A", "IT", 1500),
    ("E002", "Tran Thi B", "HR", 1200),
    ("E003", "Le Van C", "IT", 1800),
    ("E004", "Pham Thi D", "Finance", 2000),
    ("E005", "Hoang Van E", "IT", 1700)
]
YÊU CẦU: 
1.Duyệt danh sách nhân viên.? 
2.Lọc dữ liệu nv thuộc phòng IT.? 
3 Thống kê lương (Tính tổng lương, lương trung bình) của công ty.?
4.Cập nhật lương nhưng KHÔNG sửa tuple
    - Tạo list mới updated_employees
    - Nhân viên phòng IT tăng lương 10%
    - Các phòng khác giữ nguyên
    - Employees ban đầu không được thay đổi. 
    - Thêm nhân viên mới, in lại danh sách nv sau khi thêm. 



```python
employees = [
    ("E001", "Nguyen Van A", "IT", 1500),
    ("E002", "Tran Thi B", "HR", 1200),
    ("E003", "Le Van C", "IT", 1800),
    ("E004", "Pham Thi D", "Finance", 2000),
    ("E005", "Hoang Van E", "IT", 1700)
]

#1.Duyệt danh sách nhân viên
print("\n=== 1. DUYỆT DS NHÂN VIÊN ===")
for emp_id, name, dept, salary in employees:
    print(f"ID: {emp_id}, Name: {name}, Department: {dept}, Salary: {salary}")

#2.Lọc nhân viên IT có lương > 1600
print("\n=== 2. NHÂN VIÊN PHÒNG IT ===")
for emp_id, name, dept, salary in employees:
    if dept == "IT":
        print(f"{emp_id} | {name} | {salary}")

#3.THỐNG KÊ LƯƠNG NV CTY 
print("\n=== 3. TỔNG LƯƠNG VÀ LƯƠNG TRUNG BÌNH ===")
total_salary = sum(salary for emp_id, name, dept, salary in employees)
avg_salary = total_salary / len(employees) # trong python không có hàm avg nên ta tự tính 
print(f"Tổng lương: {total_salary}")
print(f"Lương trung bình: {avg_salary}")

#4. Cập nhật lương nhưng KHÔNG sửa tuple,Tạo list mới updated_employees
#IT tăng 10%
#Phòng khác giữ nguyên
#employees ban đầu không đổi

print("\n=== 4. CẬP NHẬT LƯƠNG NHÂN VIÊN ===")
updated_employees = []
for emp_id, name, dept, salary in employees:
    if dept == "IT":
        new_salary = salary * 1.10
    else:
        new_salary = salary

    # tạo tuple mới
    updated_employees.append(
        (emp_id, name, dept, new_salary)
    )

print("\n=== DANH SÁCH SAU KHI CẬP NHẬT LƯƠNG ===")
for emp in updated_employees:
    print(emp)
#5. TThêm nhân viên mới & in lại danh sách
print("\n=== 5. THÊM NHÂN VIÊN MỚI ===")
new_employee = ("E006", "Nguyễn Thị Diễm Hương", "Marketing", 1600)
updated_employees.append(new_employee)
print("\n=== DANH SÁCH SAU KHI THÊM NHÂN VIÊN MỚI ===")
for emp_id, name, dept, salary in updated_employees:
    print(f"{emp_id} | {name} | {dept} | {salary}")

```

3.Thống kê doanh số bán hàng của từng nhân viên trong một kỳ (tuần / tháng).
Theo mã:  Mã nhân viên, Danh sách doanh số các đơn hàng. 
    Định nghĩa hàm calculate_sales_stats(sales)
    Hàm nhận vào list doanh số
Trả về tuple gồm:
    Tổng doanh số
    Doanh số cao nhất
    Doanh số trung bình
    In báo cáo doanh số cho nhân viên

```python
salaries = [1500, 1200, 1800, 2000, 1700]

def salary_stats(salary_list):
    min_salary = min(salary_list)
    max_salary = max(salary_list)
    avg_salary = sum(salary_list) / len(salary_list)
    return min_salary, max_salary, avg_salary

min_s, max_s, avg_s = salary_stats(salaries)

print(f"Min: {min_s}, Max: {max_s}, Average: {avg_s}")
print("Số nhân viên:", len(salaries))
```

```python
#3.Thống kê doanh số bán hàng của từng nhân viên trong một kỳ (tuần / tháng).
# Định nghĩa hàm tính doanh số
def calculate_sales_stats(sales):
    total_sales = sum(sales)
    max_sales = max(sales)
    avg_sales = total_sales / len(sales)
    return total_sales, max_sales, avg_sales

# Dữ liệu doanh số của một nhân viên
employee_id = "E007"
monthly_sales = [1200, 1800, 1500, 2000, 1700]

# Gọi hàm đã định nghĩa
total, max_val, avg_val = calculate_sales_stats(monthly_sales)
# In báo cáo
print(f"Employee ID: {employee_id}")
print(f"Total Sales: {total}")
print(f"Max Sale: {max_val}")
print(f"Average Sales: {avg_val}")
print("Number of orders:", len(monthly_sales))
```

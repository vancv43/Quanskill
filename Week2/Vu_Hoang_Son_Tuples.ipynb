YÊU CẦU :
KỸ THUẬT ACCESS TUPLES:
1. Lọc Tên sinh viên và chuyên ngành của sinh viên "SV01".

KỸ THUẬT LOOP + UNPACK TUPLES:
2. Dùng vòng lặp for + tuple unpacking để in toàn bộ bảng điểm SV.

KỸ THUẬT UPDATE TUPLES:

KỸ THUẬT THỐNG KÊ ĐIỂM (LOOP + FUNCTION):

JOIN TUPLES (RẤT QUAN TRỌNG): # là sự kết hợp (nối 2 dữ liệu từ 2 Tupes lại với nhau)
7.Thực hiện JOIN giữa students và scores theo student_id để in báo cáo.
Output: "Nguyen Van Tuan" | Data Science | Python | 8.5




```python
#Lọc Tên sinh viên và chuyên ngành của sinh viên "SV01"
students = [ ("SV01", "Nguyen Van Tuan", "Data Science"), ("SV02", "Tran Thi Phuong", "AI"), ("SV03", "Le Van Dung", "Data Science")]

for student in students:
  if student[0]=="SV01":   #Access mã sinh viên
     name = student[1]       #Access tên sinh viên
     major = student[2]      #Access chuyên ngành
     print("Tên sinh viên:", name)
     print ("Chuyên ngành:", major)
```

```python
#KỸ THUẬT LOOP + UNPACK TUPLES: 2. Dùng vòng lặp for + tuple unpacking để in toàn bộ bảng điểm SV.
scores = [ ("SV01", "Python", 8.5), ("SV01", "Math", 7.5), ("SV01", "Marketing", 7.0), ("SV02", "Accounting", 9.0), ("SV02", "Python", 8.5), ("SV03", "Math", 7.5), ("SV03", "Marketing", 6.5) ]

for student_id, subject, score in scores:
      print(f"Mã sinh viên: {student_id} | Môn học: {subject} | Điểm: {score}")
```

```python
#KỸ THUẬT UPDATE TUPLES:
#3. Tạo danh sách mới updated_scores
#4. Sinh viên SV03 được cộng thêm 1.0 điểm Accounting
#5. Dữ liệu scores ban đầu không được thay đổi
scores = [ ("SV01", "Python", 8.5), ("SV01", "Math", 7.5), ("SV01", "Marketing", 7.0), ("SV02", "Accounting", 9.0), ("SV02", "Python", 8.5), ("SV03", "Math", 7.5), ("SV03", "Marketing", 6.5) ]

updated_score = []
for student_id, subject, score in scores:
    if student_id == "SV03":
       new_score = score + 1.0
       updated_score.append((student_id, subject, new_score))
    else:
       updated_score.append((student_id, subject, score))

for s in updated_score:
    print(s)
```

```python
#KỸ THUẬT THỐNG KÊ ĐIỂM (LOOP + FUNCTION): 6. def calculate_average(student_id, scores), tính điểm trung bình của sinh viên có mã số SV01.
scores = [ ("SV01", "Python", 8.5), ("SV01", "Math", 7.5), ("SV01", "Marketing", 7.0), ("SV02", "Accounting", 9.0), ("SV02", "Python", 8.5), ("SV03", "Math", 7.5), ("SV03", "Marketing", 6.5) ]

def calculate_average(student_id, scores):
    total = 0
    count = 0

    for sid, subject, score in scores:   #LOOP
        if sid == student_id:
            total += score
            count += 1

    return total / count if count > 0 else 0

avg_score = calculate_average(student_id, scores)
print(f"Mã sinh viên: {student_id} | Điểm trung bình: {avg_score}")

```

```python
#JOIN TUPLES (RẤT QUAN TRỌNG): # là sự kết hợp (nối 2 dữ liệu từ 2 Tupes lại với nhau) 7.Thực hiện JOIN giữa students và scores theo student_id để in báo cáo. Output: "Nguyen Van Tuan" | Data Science | Python | 8.5

students = [ ("SV01", "Nguyen Van Tuan", "Data Science"), ("SV02", "Tran Thi Phuong", "AI"), ("SV03", "Le Van Dung", "Data Science")]
scores = [ ("SV01", "Python", 8.5), ("SV01", "Math", 7.5), ("SV01", "Marketing", 7.0), ("SV02", "Accounting", 9.0), ("SV02", "Python", 8.5), ("SV03", "Math", 7.5), ("SV03", "Marketing", 6.5) ]

for student_id, name, major in students:
  for score_student_id, subject, score in scores:
    if student_id == score_student_id:
       print(f"{name} | {major} | {subject} | {score}")
```

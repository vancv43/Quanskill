## Giới thiệu về câu lệnh điều kiện

Chào mọi người! Trong bài giảng này, chúng ta sẽ tìm hiểu về **câu lệnh điều kiện** (conditionals) và cách chúng ta có thể kiểm soát việc thực thi các câu lệnh dựa trên việc một giá trị có được đánh giá là `True` (đúng) hay `False` (sai). Những giá trị `True` và `False` này được gọi là **kiểu Boolean**. Hãy bắt đầu với một ví dụ đơn giản nhất về câu lệnh điều kiện: câu lệnh `if`.

### Câu lệnh `if` cơ bản

Hãy viết một câu lệnh `if` đơn giản để kiểm tra một điều kiện. Ví dụ, chúng ta sẽ kiểm tra xem một điều kiện có đánh giá là `True` hay không:

````python
if True:
    print("Điều kiện đúng!")
````

**Giải thích**:  
- Ở đây, chúng ta sử dụng từ khóa `if` theo sau là điều kiện `True`.  
- Dấu hai chấm `:` đánh dấu bắt đầu của khối mã bên trong câu lệnh `if`.  
- Khối mã bên trong phải được **thụt lề** (indent) để Python hiểu rằng nó thuộc về câu lệnh `if`.  
- Khi chạy đoạn code này, dòng chữ `"Điều kiện đúng!"` sẽ được in ra vì điều kiện `True` được thỏa mãn.  

Bây giờ, nếu chúng ta đổi điều kiện thành `False`:

````python
if False:
    print("Điều kiện đúng!")
````

Khi chạy đoạn code này, không có gì được in ra vì điều kiện `False` không thỏa mãn, và khối mã bên trong câu lệnh `if` không được thực thi.

### Sử dụng biến trong câu lệnh điều kiện

Trong thực tế, chúng ta hiếm khi sử dụng giá trị `True` hoặc `False` cứng (hard-coded). Thay vào đó, chúng ta sử dụng các biểu thức hoặc biến mà khi đánh giá sẽ trả về `True` hoặc `False`. Hãy thử một ví dụ:

````python
language = "Python"
if language == "Python":
    print("Ngôn ngữ là Python!")
````

**Giải thích**:  
- Chúng ta tạo một biến `language` và gán giá trị `"Python"` cho nó.  
- Trong câu lệnh `if`, chúng ta sử dụng toán tử so sánh `==` (double equals) để kiểm tra xem giá trị của biến `language` có bằng `"Python"` hay không.  
- Toán tử `==` kiểm tra sự **bằng nhau** (equality), khác với toán tử `=` dùng để gán giá trị.  
- Khi chạy code, dòng chữ `"Ngôn ngữ là Python!"` sẽ được in ra vì điều kiện `language == "Python"` là `True`.

### Các toán tử so sánh

Dưới đây là danh sách các toán tử so sánh mà chúng ta có thể sử dụng trong Python:

- `==`: Kiểm tra bằng nhau (equality).  
- `!=`: Kiểm tra không bằng nhau (inequality).  
- `>`: Lớn hơn.  
- `<`: Nhỏ hơn.  
- `>=`: Lớn hơn hoặc bằng.  
- `<=`: Nhỏ hơn hoặc bằng.  
- `is`: Kiểm tra danh tính đối tượng (object identity).  

Ví dụ về các toán tử so sánh:

````python
a = 5
b = 10
if a == b:
    print("a bằng b")
if a != b:
    print("a không bằng b")
if a < b:
    print("a nhỏ hơn b")
````

**Kết quả khi chạy**:  
```
a không bằng b
a nhỏ hơn b
```

### Câu lệnh `else`

Nếu chúng ta muốn thực thi một đoạn mã khi điều kiện trong câu lệnh `if` không thỏa mãn, chúng ta có thể sử dụng câu lệnh `else`. Hãy mở rộng ví dụ trước:

````python
language = "Python"
if language == "Python":
    print("Ngôn ngữ là Python!")
else:
    print("Không khớp!")
````

**Giải thích**:  
- Nếu `language == "Python"` là `True`, dòng `"Ngôn ngữ là Python!"` sẽ được in.  
- Nếu điều kiện là `False` (ví dụ, thay `language = "Java"`), dòng `"Không khớp!"` sẽ được in.  
- Lưu ý: Câu lệnh `else` phải được đặt ở cùng mức thụt lề với câu lệnh `if` (không nằm trong khối `if`).

Hãy thử thay đổi giá trị của `language`:

````python
language = "Java"
if language == "Python":
    print("Ngôn ngữ là Python!")
else:
    print("Không khớp!")
````

**Kết quả khi chạy**:  
```
Không khớp!
```

### Câu lệnh `elif`

Nếu chúng ta muốn kiểm tra nhiều điều kiện khác nhau, chúng ta có thể sử dụng câu lệnh `elif` (viết tắt của "else if"). Ví dụ, kiểm tra nhiều ngôn ngữ lập trình:

````python
language = "Java"
if language == "Python":
    print("Ngôn ngữ là Python!")
elif language == "Java":
    print("Ngôn ngữ là Java!")
elif language == "JavaScript":
    print("Ngôn ngữ là JavaScript!")
else:
    print("Không khớp!")
````

**Giải thích**:  
- Python sẽ kiểm tra từng điều kiện theo thứ tự:  
  1. Nếu `language == "Python"` là `True`, in `"Ngôn ngữ là Python!"`.  
  2. Nếu không, kiểm tra `language == "Java"`. Nếu đúng, in `"Ngôn ngữ là Java!"`.  
  3. Nếu không, kiểm tra `language == "JavaScript"`. Nếu đúng, in `"Ngôn ngữ là JavaScript!"`.  
  4. Nếu không điều kiện nào thỏa mãn, in `"Không khớp!"`.  
- Khi chạy đoạn code trên với `language = "Java"`, kết quả sẽ là:  
```
Ngôn ngữ là Java!
```

**Lưu ý**: Python không có câu lệnh `switch-case` như một số ngôn ngữ khác, nhưng các câu lệnh `if`, `elif`, và `else` đủ mạnh mẽ và rõ ràng để thay thế.

### Toán tử Boolean: `and`, `or`, `not`

Ngoài các toán tử so sánh, chúng ta còn có các **toán tử Boolean** để kết hợp nhiều điều kiện:

- `and`: Yêu cầu cả hai điều kiện đều phải là `True`.  
- `or`: Chỉ cần một trong hai điều kiện là `True`.  
- `not`: Đảo ngược giá trị Boolean (`True` thành `False` và ngược lại).

#### Ví dụ với `and`

````python
user = "admin"
logged_in = True
if user == "admin" and logged_in:
    print("Trang quản trị")
else:
    print("Thông tin xác thực không hợp lệ")
````

**Giải thích**:  
- Điều kiện `user == "admin" and logged_in` chỉ là `True` nếu cả `user == "admin"` và `logged_in == True`.  
- Khi chạy code trên, kết quả là:  
```
Trang quản trị
```

Nếu thay đổi `logged_in = False`:

````python
user = "admin"
logged_in = False
if user == "admin" and logged_in:
    print("Trang quản trị")
else:
    print("Thông tin xác thực không hợp lệ")
````

**Kết quả**:  
```
Thông tin xác thực không hợp lệ
```

#### Ví dụ với `or`

````python
user = "admin"
logged_in = False
if user == "admin" or logged_in:
    print("Trang quản trị")
else:
    print("Thông tin xác thực không hợp lệ")
````

**Giải thích**:  
- Với toán tử `or`, chỉ cần một trong hai điều kiện là `True`. Ở đây, `user == "admin"` là `True`, nên kết quả là:  
```
Trang quản trị
```

#### Ví dụ với `not`

````python
logged_in = False
if not logged_in:
    print("Vui lòng đăng nhập")
else:
    print("Chào mừng bạn!")
````

**Giải thích**:  
- Toán tử `not` đảo ngược giá trị của `logged_in`. Vì `logged_in = False`, nên `not logged_in` là `True`.  
- Kết quả khi chạy:  
```
Vui lòng đăng nhập
```

### Toán tử `is` và danh tính đối tượng

Toán tử `==` kiểm tra giá trị bằng nhau, còn toán tử `is` kiểm tra xem hai đối tượng có cùng **danh tính** (identity), tức là cùng một đối tượng trong bộ nhớ hay không.

Ví dụ:

````python
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)  # Kiểm tra giá trị
print(a is b)  # Kiểm tra danh tính
print(id(a))   # In địa chỉ bộ nhớ của a
print(id(b))   # In địa chỉ bộ nhớ của b
````

**Kết quả**:  
```
True
False
140735123456789  # Địa chỉ của a
140735987654321  # Địa chỉ của b
```

**Giải thích**:  
- `a == b` trả về `True` vì hai danh sách có giá trị giống nhau.  
- `a is b` trả về `False` vì `a` và `b` là hai đối tượng khác nhau trong bộ nhớ (có ID khác nhau).  

Nếu gán `b = a`:

````python
a = [1, 2, 3]
b = a
print(a == b)  # Kiểm tra giá trị
print(a is b)  # Kiểm tra danh tính
print(id(a))   # In địa chỉ bộ nhớ của a
print(id(b))   # In địa chỉ bộ nhớ của b
````

**Kết quả**:  
```
True
True
140735123456789
140735123456789
```

**Giải thích**:  
- Bây giờ `b = a` nghĩa là `b` tham chiếu đến cùng đối tượng với `a` trong bộ nhớ. Do đó, `a is b` trả về `True`.

### Các giá trị đánh giá là `False` trong Python

Để sử dụng câu lệnh điều kiện hiệu quả, chúng ta cần hiểu những giá trị nào trong Python được đánh giá là `True` hoặc `False`. Dưới đây là các giá trị đánh giá là `False`:

1. Giá trị `False`.  
2. Giá trị `None`.  
3. Số `0` (bao gồm `0.0`, `0j`).  
4. Chuỗi rỗng `""`, danh sách rỗng `[]`, tuple rỗng `()`, từ điển rỗng `{}`.  

Mọi giá trị khác đều được đánh giá là `True`. Hãy xem ví dụ:

````python
condition = False
if condition:
    print("Đánh giá là True")
else:
    print("Đánh giá là False")
````

**Kết quả**:  
```
Đánh giá là False
```

Thử với `None`:

````python
condition = None
if condition:
    print("Đánh giá là True")
else:
    print("Đánh giá là False")
````

**Kết quả**:  
```
Đánh giá là False
```

Thử với số `0`:

````python
condition = 0
if condition:
    print("Đánh giá là True")
else:
    print("Đánh giá là False")
````

**Kết quả**:  
```
Đánh giá là False
```

Thử với danh sách rỗng:

````python
condition = []
if condition:
    print("Đánh giá là True")
else:
    print("Đánh giá là False")
````

**Kết quả**:  
```
Đánh giá là False
```

Thử với một chuỗi không rỗng:

````python
condition = "test"
if condition:
    print("Đánh giá là True")
else:
    print("Đánh giá là False")
````

**Kết quả**:  
```
Đánh giá là True
```

**Ứng dụng**:  
Việc biết các giá trị đánh giá là `False` rất hữu ích khi kiểm tra xem một chuỗi, danh sách, hoặc từ điển có rỗng hay không mà không cần gọi phương thức cụ thể.

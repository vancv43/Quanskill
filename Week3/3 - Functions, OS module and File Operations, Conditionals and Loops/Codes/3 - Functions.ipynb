# Hàm (Functions) trong Python

---

Xin chào mọi người! Trong bài giảng này, chúng ta sẽ tìm hiểu về **hàm** trong Python. Hàm là một tập hợp các câu lệnh được đóng gói lại với nhau để thực hiện một nhiệm vụ cụ thể. Hàm giúp chúng ta tổ chức mã nguồn tốt hơn, dễ bảo trì và tái sử dụng. Hãy cùng bắt đầu bằng cách tạo một hàm đơn giản để thấy được lợi ích của nó.

---

## Tạo hàm đầu tiên

Để tạo một hàm trong Python, chúng ta sử dụng từ khóa **`def`** (viết tắt của "definition" - định nghĩa). Hãy tạo một hàm đơn giản có tên là `hello_func`:

````
def hello_func():
    pass
````

### Giải thích:
- **`def`**: Từ khóa để định nghĩa hàm.
- **`hello_func`**: Tên của hàm. Tên hàm nên rõ ràng, mô tả chức năng của nó.
- **Dấu ngoặc `()`**: Đây là nơi chúng ta đặt các tham số (parameters) của hàm. Hiện tại, hàm này chưa có tham số nên để trống.
- **`pass`**: Từ khóa này được sử dụng khi chúng ta muốn định nghĩa hàm nhưng chưa muốn viết mã bên trong. Nó giúp tránh lỗi cú pháp khi để hàm trống.

Nếu chúng ta muốn gọi (thực thi) hàm này, chỉ cần viết:

````
hello_func()
````

**Lưu ý**: Khi gọi hàm, chúng ta phải thêm dấu ngoặc `()` để thực thi nó. Nếu không có dấu ngoặc, Python sẽ chỉ tham chiếu đến hàm mà không chạy nó. Ví dụ:

````
print(hello_func)  # In ra: <function hello_func at ...>
````

Kết quả này cho thấy `hello_func` là một đối tượng hàm trong bộ nhớ, nhưng nó không được thực thi. Để thực thi, ta cần:

````
hello_func()  # Hiện tại không in gì vì hàm chỉ có pass
````

Bây giờ, hãy thêm một câu lệnh in vào hàm:

````
def hello_func():
    print("Hello Function!")

hello_func()  # Kết quả: Hello Function!
````

Khi gọi `hello_func()`, hàm sẽ in ra chuỗi `"Hello Function!"`.

---

## Lợi ích của hàm: Tái sử dụng mã

Một trong những lợi ích lớn nhất của hàm là giúp chúng ta **tái sử dụng mã** mà không cần lặp lại. Hãy xem một ví dụ: Giả sử bạn cần in chuỗi `"Hello Function!"` ở nhiều nơi trong chương trình:

````
print("Hello Function!")
print("Hello Function!")
print("Hello Function!")
print("Hello Function!")
````

Nếu sếp yêu cầu đổi dấu `!` thành dấu `.`, bạn sẽ phải sửa từng dòng một. Điều này không hiệu quả, đặc biệt nếu mã nguồn lớn và chuỗi này xuất hiện ở hàng trăm nơi.

Thay vào đó, nếu sử dụng hàm:

````
def hello_func():
    print("Hello Function!")

hello_func()
hello_func()
hello_func()
hello_func()
````

Giờ đây, nếu cần đổi dấu `!` thành dấu `.`, bạn chỉ cần sửa trong hàm:

````
def hello_func():
    print("Hello Function.")

hello_func()
hello_func()
hello_func()
hello_func()
````

Kết quả sẽ thay đổi ở tất cả các nơi gọi hàm, giúp mã nguồn **DRY** (Don't Repeat Yourself - Không lặp lại chính mình). Đây là nguyên tắc quan trọng trong lập trình.

## Trả về giá trị từ hàm

Hàm trở nên mạnh mẽ hơn khi chúng ta sử dụng từ khóa **`return`** để trả về một giá trị. Thay vì in chuỗi, hãy để hàm trả về chuỗi:

````
def hello_func():
    return "Hello Function."

print(hello_func())  # Kết quả: Hello Function.
````

### Giải thích:
- Khi hàm trả về một giá trị (ở đây là chuỗi `"Hello Function."`), giá trị này có thể được sử dụng như một biến thông thường.
- Ví dụ,_featured

Hàm là một "cỗ máy" nhận đầu vào (input) và tạo ra đầu ra (output). Bạn không cần biết chi tiết cách hàm hoạt động, chỉ cần quan tâm đến đầu vào và đầu ra.

Ví dụ, chúng ta có thể sử dụng giá trị trả về của hàm như một chuỗi:

````
print(hello_func().upper())  # Kết quả: HELLO FUNCTION.
````

Ở đây, giá trị trả về là chuỗi `"Hello Function."`, và chúng ta có thể gọi phương thức chuỗi `.upper()` để chuyển thành chữ in hoa.

## Truyền tham số vào hàm

Hàm có thể nhận **tham số** (parameters) để xử lý dữ liệu đầu vào. Hãy sửa hàm `hello_func` để nhận một tham số `greeting`:

````
def hello_func(greeting):
    return "{} Function.".format(greeting)

print(hello_func("Hi"))  # Kết quả: Hi Function.
````

Nếu bạn gọi hàm mà không truyền tham số, Python sẽ báo lỗi:

````
hello_func()  # Lỗi: missing 1 required positional argument: 'greeting'
````

---

## Tham số mặc định

Chúng ta có thể gán giá trị mặc định cho tham số để không bắt buộc phải truyền giá trị:

````
def hello_func(greeting, name="You"):
    return "{}, {}.".format(greeting, name)

print(hello_func("Hi"))  # Kết quả: Hi, You.
print(hello_func("Hi", "Corey"))  # Kết quả: Hi, Corey.
````

### Giải thích:
- Tham số `name` có giá trị mặc định là `"You"`. Nếu không truyền giá trị cho `name`, hàm sẽ sử dụng giá trị mặc định.
- Tham số bắt buộc (như `greeting`) phải đứng trước tham số mặc định (như `name`) trong định nghĩa hàm.

## Tham số *args và **kwargs

Python cho phép hàm nhận số lượng tham số không xác định bằng cách sử dụng `*args` (cho tham số vị trí) và `**kwargs` (cho tham số từ khóa). Ví dụ:

````
def student_info(*args, **kwargs):
    print(args)  # Tuple chứa các tham số vị trí
    print(kwargs)  # Dictionary chứa các tham số từ khóa

student_info("Math", "Art", name="John", age=22)
````

Kết quả:
```
('Math', 'Art')
{'name': 'John', 'age': 22}
```

### Giải thích:
- `*args`: Thu thập tất cả các tham số vị trí vào một **tuple**.
- `**kwargs`: Thu thập tất cả các tham số từ khóa vào một **dictionary**.

Chúng ta cũng có thể sử dụng `*` và `**` để "giải nén" danh sách hoặc dictionary:

````
courses = ["Math", "Art"]
info = {"name": "John", "age": 22}

student_info(*courses, **info)
````

Kết quả tương tự như trên, vì `*courses` giải nén danh sách thành các tham số vị trí riêng lẻ, và `**info` giải nén dictionary thành các tham số từ khóa.

## Ví dụ thực tế: Hàm từ thư viện chuẩn Python

Hãy xem một ví dụ thực tế với hai hàm từ thư viện chuẩn Python, được điều chỉnh nhẹ:

````
month_days = [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

def is_leap(year):
    """Return True for leap years, False for non-leap years."""
    return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)

def days_in_month(year, month):
    """Return number of days in that month in that year."""
    if not 1 <= month <= 12:
        return "Invalid month"
    if month == 2 and is_leap(year):
        return 29
    return month_days[month]
````

### Giải thích:
- **`month_days`**: Danh sách chứa số ngày của từng tháng (tháng 0 là placeholder).
- **`is_leap(year)`**: Kiểm tra xem một năm có phải là năm nhuận hay không.
  - Điều kiện: Năm chia hết cho 4, không chia hết cho 100, hoặc chia hết cho 400.
- **`days_in_month(year, month)`**: Trả về số ngày trong tháng của năm được chỉ định.
  - Nếu tháng không hợp lệ (ngoài khoảng 1-12), trả về thông báo lỗi.
  - Nếu là tháng 2 và năm nhuận, trả về 29 ngày.
  - Nếu không, trả về số ngày từ danh sách `month_days`.

Hãy thử chạy:

````
print(is_leap(2017))  # Kết quả: False
print(is_leap(2020))  # Kết quả: True
print(days_in_month(2017, 2))  # Kết quả: 28
````

### Phân tích cách hoạt động của `days_in_month(2017, 2)`:
1. Hàm nhận `year = 2017` và `month = 2`.
2. Kiểm tra: `month = 2` nằm trong khoảng 1-12, nên hợp lệ.
3. Kiểm tra: `month == 2` là đúng, nhưng `is_leap(2017)` trả về `False` (2017 không phải năm nhuận).
4. Vì vậy, hàm trả về `month_days[2] = 28`.

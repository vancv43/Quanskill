## Giới Thiệu về Module OS

Chào mọi người! Trong bài giảng này, chúng ta sẽ tìm hiểu về **module `os`** trong Python và một số chức năng hữu ích mà nó cung cấp. Module `os` cho phép chúng ta tương tác với hệ điều hành cơ bản theo nhiều cách khác nhau. Ví dụ, chúng ta có thể:

- Điều hướng hệ thống tệp (file system).

- Thay đổi hoặc xem các biến môi trường (environment variables).

## Bắt Đầu với Module OS

Đầu tiên, để sử dụng module `os`, chúng ta cần nhập nó vào chương trình:

````python
import os
````

### Xem Tất Cả Chức Năng của Module OS

Khi làm việc với một module mới, một mẹo hữu ích là sử dụng hàm `dir()` để liệt kê tất cả các thuộc tính và phương thức có sẵn trong module. Hãy thử chạy đoạn mã sau:

````python
print(dir(os))
````

Kết quả sẽ hiển thị một danh sách dài các phương thức và thuộc tính của module `os`. Trong bài giảng này, chúng ta sẽ không đi qua từng phương thức, mà chỉ tập trung vào những chức năng mà tôi thường sử dụng và được xem là hữu ích nhất.

## 1. Lấy Thư Mục Làm Việc Hiện Tại

Một trong những chức năng cơ bản của module `os` là lấy thư mục làm việc hiện tại (current working directory). Chúng ta sử dụng phương thức `os.getcwd()`:

````python
import os

print(os.getcwd())
````

**Kết quả**: In ra đường dẫn của thư mục hiện tại, ví dụ: `/Users/username/Desktop/module_os`.

Điều này hữu ích khi bạn muốn biết chương trình đang chạy trong thư mục nào.

---

## 2. Thay Đổi Thư Mục Làm Việc

Nếu bạn muốn chuyển sang một thư mục khác trên hệ thống tệp, bạn có thể sử dụng phương thức `os.chdir()` (change directory). Ví dụ, để chuyển đến thư mục Desktop:

````python
import os

# Thay đổi thư mục làm việc sang Desktop
os.chdir('/users/macbook/Desktop')   

# In thư mục làm việc hiện tại sau khi thay đổi
print(os.getcwd())
````

**Kết quả**: 
```
/Users/username/Desktop
```

Sau khi chạy đoạn mã trên, thư mục làm việc hiện tại sẽ chuyển sang Desktop.

---

## 3. Liệt Kê Tệp và Thư Mục

Để xem danh sách các tệp và thư mục trong thư mục hiện tại, bạn có thể sử dụng phương thức `os.listdir()`. Phương thức này trả về một danh sách các tệp và thư mục trong thư mục được chỉ định (mặc định là thư mục hiện tại):

````python
import os

# Liệt kê các tệp và thư mục trên Desktop
print(os.listdir())
````

**Kết quả**: 
```
['file1.txt', 'folder1', 'demo.txt', 'module_os']
```

Bạn cũng có thể truyền vào một đường dẫn cụ thể nếu muốn liệt kê nội dung của thư mục khác:

````python
print(os.listdir('/Users/username/Desktop'))
````

## 4. Tạo Thư Mục Mới

Module `os` cung cấp hai phương thức để tạo thư mục: `os.mkdir()` và `os.makedirs()`. Cả hai đều tạo thư mục mới, nhưng có một số khác biệt quan trọng.

### Sử dụng `os.mkdir()`

Phương thức `os.mkdir()` tạo một thư mục đơn cấp (single-level directory). Ví dụ:

````python
import os

# Tạo thư mục mới có tên 'os-demo-2'
os.mkdir('os-demo-2')
````

Nếu thư mục mẹ (parent directory) không tồn tại, `os.mkdir()` sẽ báo lỗi.

### Sử dụng `os.makedirs()`

Phương thức `os.makedirs()` tạo thư mục một cách đệ quy, tức là nó sẽ tạo cả các thư mục trung gian nếu cần. Ví dụ:

````python
import os

# Tạo thư mục 'os-demo-2/subdir1' (tạo cả thư mục mẹ nếu chưa tồn tại)
os.makedirs('os-demo-2/subdir1')
````

**Kết quả**: Thư mục `os-demo-2` được tạo, bên trong chứa thư mục con `subdir1`.

**Lưu ý**: Tôi thường sử dụng `os.makedirs()` ngay cả khi chỉ tạo thư mục cấp cao nhất, vì nó linh hoạt hơn và có thể xử lý các trường hợp cần tạo cấu trúc thư mục phức tạp.

---

## 5. Xóa Thư Mục

Tương tự như tạo thư mục, module `os` cung cấp hai phương thức để xóa thư mục: `os.rmdir()` và `os.removedirs()`.

### Sử dụng `os.rmdir()`

Phương thức `os.rmdir()` xóa một thư mục cụ thể, nhưng thư mục đó phải trống:

````python
import os

# Xóa thư mục 'os-demo-2/subdir1'
os.rmdir('os-demo-2/subdir1')
````

### Sử dụng `os.removedirs()`

Phương thức `os.removedirs()` xóa thư mục một cách đệ quy, tức là xóa cả thư mục mẹ nếu nó trở nên trống sau khi xóa thư mục con:

````python
import os

# Xóa toàn bộ cây thư mục 'os-demo-2'
os.removedirs('os-demo-2/subdir1')
````

**Lưu ý**: Việc xóa thư mục đệ quy có thể nguy hiểm, vì vậy tôi thường sử dụng `os.rmdir()` để kiểm soát chính xác thư mục cần xóa.

---

## 6. Đổi Tên Tệp hoặc Thư Mục

Để đổi tên một tệp hoặc thư mục, bạn có thể sử dụng phương thức `os.rename()`. Phương thức này nhận hai đối số: tên hiện tại và tên mới.

Ví dụ, để đổi tên tệp `test.txt` thành `demo.txt`:

````python
import os

# Đổi tên tệp từ 'test.txt' thành 'demo.txt'
os.rename('test.txt', 'demo.txt')

# Kiểm tra danh sách tệp sau khi đổi tên
print(os.listdir())
````

**Kết quả**: 
```
['file1.txt', 'folder1', 'demo.txt', 'module_os']
```

Tệp `test.txt` đã được đổi tên thành `demo.txt`.

---

## 7. Lấy Thông Tin về Tệp

Để lấy thông tin chi tiết về một tệp, bạn có thể sử dụng phương thức `os.stat()`. Phương thức này trả về một đối tượng chứa thông tin như kích thước tệp, thời gian sửa đổi, v.v.

Ví dụ:

````python
import os

# Lấy thông tin về tệp 'demo.txt'
print(os.stat('demo.txt'))
````

**Kết quả**: 
```
os.stat_result(st_mode=33188, st_ino=123456, st_dev=16777220, st_nlink=1, st_uid=501, st_gid=20, st_size=20, st_atime=1634567890, st_mtime=1634567890, st_ctime=1634567890)
```

Một số thuộc tính hữu ích:
- `st_size`: Kích thước tệp (tính bằng byte).
- `st_mtime`: Thời gian sửa đổi lần cuối (last modification time).

### In Kích Thước Tệp

````python
import os

# Lấy kích thước tệp 'demo.txt'
file_info = os.stat('demo.txt')
print(file_info.st_size)  # Kích thước: 20 bytes
````

### In Thời Gian Sửa Đổi ở Định Dạng Dễ Đọc

Thời gian sửa đổi (`st_mtime`) được trả về dưới dạng timestamp. Để chuyển đổi thành định dạng dễ đọc, chúng ta sử dụng module `datetime`:

````python
import os
from datetime import datetime

# Lấy thời gian sửa đổi của tệp 'demo.txt'
file_info = os.stat('demo.txt')
mod_time = file_info.st_mtime
print(datetime.fromtimestamp(mod_time))
````

**Kết quả**: 
```
2021-10-18 12:34:50
```

Điều này rất hữu ích khi làm việc với các ứng dụng web cần theo dõi thời gian tạo hoặc cập nhật tệp.

## 8. Duyệt Cây Thư Mục với `os.walk()`

Phương thức `os.walk()` là một công cụ mạnh mẽ để duyệt qua toàn bộ cây thư mục (directory tree). Nó là một generator, trả về một bộ ba giá trị cho mỗi thư mục mà nó duyệt qua:

1. **Đường dẫn thư mục** (directory path).
2. **Danh sách các thư mục con** trong thư mục đó.
3. **Danh sách các tệp** trong thư mục đó.

Ví dụ, để duyệt toàn bộ cây thư mục từ Desktop:

````python
import os

# Duyệt cây thư mục từ Desktop
for dir_path, dir_names, file_names in os.walk('/Users/username/Desktop'):
    print('Đường dẫn:', dir_path)
    print('Thư mục:', dir_names)
    print('Tệp:', file_names)
    print()
````

**Kết quả** (giả định):
```
Đường dẫn: /Users/username/Desktop
Thư mục: ['folder1', 'module_os']
Tệp: ['demo.txt', 'file1.txt']

Đường dẫn: /Users/username/Desktop/folder1
Thư mục: ['subdir1']
Tệp: ['file2.txt']

Đường dẫn: /Users/username/Desktop/folder1/subdir1
Thư mục: []
Tệp: ['file3.txt']
```

Phương thức `os.walk()` rất hữu ích khi bạn cần tìm kiếm tệp trong một cấu trúc thư mục phức tạp hoặc thu thập thông tin về tất cả các tệp trong một ứng dụng web.

## 9. Lấy Biến Môi Trường

Module `os` cho phép truy cập các biến môi trường thông qua `os.environ`. Ví dụ, để lấy đường dẫn thư mục Home của người dùng:

**Kết quả**:

## 10. Làm Việc với Đường Dẫn bằng `os.path`

Module `os.path` cung cấp nhiều phương thức hữu ích để xử lý đường dẫn tệp một cách an toàn và chính xác.

### Kết Hợp Đường Dẫn với `os.path.join()`

Thay vì nối chuỗi thủ công (có thể gây lỗi do dấu gạch chéo `/` hoặc `\`), bạn nên sử dụng `os.path.join()` để tạo đường dẫn:

````python
import os

# Lấy đường dẫn thư mục Home
home_dir = os.environ['HOME']

# Kết hợp đường dẫn với tên tệp
file_path = os.path.join(home_dir, 'test.txt')
print(file_path)
````

**Kết quả**: 
```
/Users/username/test.txt
```

Phương thức này đảm bảo đường dẫn được tạo đúng trên mọi hệ điều hành.

### Lấy Tên Tệp hoặc Thư Mục

- **`os.path.basename()`**: Lấy tên tệp từ đường dẫn.
- **`os.path.dirname()`**: Lấy tên thư mục từ đường dẫn.
- **`os.path.split()`**: Tách đường dẫn thành thư mục và tên tệp.

Ví dụ:

````python
import os

path = '/Users/username/temp/test.txt'

# Lấy tên tệp
print(os.path.basename(path))  # test.txt

# Lấy tên thư mục
print(os.path.dirname(path))   # /Users/username/temp

# Tách cả hai
print(os.path.split(path))     # ('/Users/username/temp', 'test.txt')
````

### Kiểm Tra Sự Tồn Tại của Đường Dẫn

Sử dụng `os.path.exists()` để kiểm tra xem một đường dẫn có tồn tại không:

````python
import os

path = '/Users/username/temp/test.txt'
print(os.path.exists(path))  # False (nếu đường dẫn không tồn tại)
````

### Kiểm Tra Tệp hoặc Thư Mục

- **`os.path.isdir()`**: Kiểm tra xem đường dẫn có phải là thư mục không.
- **`os.path.isfile()`**: Kiểm tra xem đường dẫn có phải là tệp không.

Ví dụ:

````python
import os

path = '/Users/username/Desktop/demo.txt'
print(os.path.isfile(path))  # True (nếu là tệp)
print(os.path.isdir(path))   # False
````

### Tách Phần Mở Rộng Tệp

Phương thức `os.path.splitext()` tách tên tệp thành phần gốc (root) và phần mở rộng (extension):

````python
import os

path = '/Users/username/temp/test.txt'
root, ext = os.path.splitext(path)
print(root)  # /Users/username/temp/test
print(ext)   # .txt
````

Phương thức này rất hữu ích khi bạn cần xử lý tên tệp mà không cần phần mở rộng.

Tương tự như module `os`, bạn có thể sử dụng `dir()` để xem tất cả các phương thức của `os.path`:

Kết quả sẽ hiển thị các phương thức như `basename`, `dirname`, `exists`, `isfile`, `isdir`, `splitext`, v.v.

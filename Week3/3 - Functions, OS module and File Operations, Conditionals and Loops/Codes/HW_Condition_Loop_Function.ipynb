**Yêu cầu thự hiện**

Câu 1.Viết hàm tính lương thực nhận (net salary) cho mỗi nhân viên, trong đó kết hợp các điều kiện True / False và các toán tử logic and, or, not.

Câu 2. Viết hàm phân loại nhân viên thành các nhóm:
    Excellent, Good, Normal, Risk dựa trên KPI, số ngày đi trễ và trạng thái vi phạm.

Câu 3.Sử dụng vòng lặp for, duyệt toàn bộ danh sách nhân viên để:

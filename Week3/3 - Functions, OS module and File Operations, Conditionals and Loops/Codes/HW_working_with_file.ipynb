```python
def update_salary_all(path, rate):
    updated_lines = []

    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()

            # ⚠️ BỎ QUA DÒNG TRỐNG
            if not line:
                continue

            parts = line.split(",")

            # ⚠️ KIỂM TRA ĐÚNG 4 CỘT
            if len(parts) != 4:
                print("Dòng lỗi, bỏ qua:", line)
                continue

            emp_id, name, salary, leave = parts

            salary = int(salary.strip())
            leave  = int(leave.strip())

            salary = int(salary * (1 + rate))

            updated_lines.append(
                f"{emp_id},{name},{salary},{leave}\n"
            )

    with open(path, "w", encoding="utf-8") as f:
        f.writelines(updated_lines)

update_salary_all("employees.txt", 0.10)
print("✔ Đã cập nhật lương toàn bộ NV")
```
